RANDOM FOREST CLASSIFIER

```python
# ## Imports
# Import necessary libraries for data processing, modeling, visualization, memory management, and Excel file handling.
import matplotlib.pyplot as plt  # For creating plots
import seaborn as sns           # For enhanced visualizations with better aesthetics
import numpy as np              # For numerical operations and array handling
from datasets import load_dataset  # To load the CoNLL-2003 dataset
from sklearn.ensemble import RandomForestClassifier  # Random Forest model for classification
from sklearn.metrics import classification_report, confusion_matrix, precision_recall_curve, auc  # Evaluation metrics
from sklearn.preprocessing import LabelEncoder  # For encoding string labels to numeric
from sklearn.feature_extraction import DictVectorizer  # For converting feature dictionaries to matrices
from scipy import sparse         # For sparse matrix handling
import gc                       # For garbage collection to manage memory
from collections import Counter # For counting NER tag occurrences
import os                       # For directory handling and file saving
import pandas as pd             # For saving classification reports as Excel files

# ## Initial Setup
# Configure Seaborn for consistent and visually appealing plots.
sns.set(style="whitegrid", palette="muted")  # Set white grid background and muted color palette

# Enable garbage collection to manage memory usage.
gc.enable()  # Turn on garbage collection

# ## Plot Directory Setup
# Define a fixed directory for saving all plots and reports.
PLOT_DIR = "plots2/randomForest"       # Directory path for Random Forest plots and reports
os.makedirs(PLOT_DIR, exist_ok=True)  # Create directory if it doesn't exist

# ## Step 1: Load CoNLL-2003 Dataset with Streaming
# Load the CoNLL-2003 dataset in streaming mode and extract NER tag names.
conll2003 = load_dataset("conll2003", streaming=True)  # Load dataset with streaming for memory efficiency
label_names = load_dataset("conll2003")["train"].features["ner_tags"].feature.names  # Get NER tag names (e.g., "O", "B-PER")

# ## Step 2: Optimized Data Preparation with Generator
# Define a generator function to stream data from the dataset.
def prepare_data_generator(dataset_split):
    for example in dataset_split:
        words = example['tokens']
        ner_tags = [label_names[tag] for tag in example['ner_tags']]
        yield {'words': words, 'tags': ner_tags}

# ### Enhanced Feature Extraction
def extract_features(words):
    features = []
    for i, w in enumerate(words):
        feat = {
            'word': w.lower(),
            'len': len(w),
            'is_upper': w.isupper(),
            'is_digit': w.isdigit(),
            'prev_word': w.lower() if i > 0 else '<START>',
            'next_word': words[i+1].lower() if i < len(words)-1 else '<END>'
        }
        features.append(feat)
    return features

# ### Process Data in Chunks
def process_data_in_chunks(split_name, chunk_size=1000):
    X_feats, y = [], []
    generator = prepare_data_generator(conll2003[split_name])
    chunk_X, chunk_y = [], []
    count = 0
    
    for data in generator:
        chunk_X.append(extract_features(data['words']))
        chunk_y.append(data['tags'])
        count += 1
        
        if count >= chunk_size:
            X_feats.extend(chunk_X)
            y.extend(chunk_y)
            chunk_X, chunk_y = [], []
            count = 0
    
    if chunk_X:
        X_feats.extend(chunk_X)
        y.extend(chunk_y)
    
    return X_feats, y

# Prepare datasets for training, validation, and test splits.
X_train_feats, y_train = process_data_in_chunks('train')
X_val_feats, y_val = process_data_in_chunks('validation')
X_test_feats, y_test = process_data_in_chunks('test')

# ## Step 3: Enhanced NER Distribution Plot with Statistics and Excel Export
def plot_ner_distribution(y_data, title):
    label_counts = Counter(tag for seq in y_data for tag in seq)
    
    # Print the statistics
    print(f"\nNER Tag Distribution Statistics for {title}:")
    for tag, count in label_counts.items():
        print(f"{tag}: {count}")
    
    # Export to Excel
    counts_df = pd.DataFrame.from_dict(label_counts, orient='index', columns=['Count'])
    counts_df.index.name = 'NER Tag'
    counts_file_name = f"RandomForest_{title.replace(' ', '_').replace(':', '')}_Counts.xlsx"
    counts_df.to_excel(f"{PLOT_DIR}/{counts_file_name}")
    
    # Create the bar plot
    plt.figure(figsize=(12, 6))
    bars = sns.barplot(
        x=list(label_counts.keys()),
        y=list(label_counts.values()),
        hue=list(label_counts.keys()),
        palette="viridis",
        legend=False,
        edgecolor="black",
        linewidth=1.5
    )
    
    # Add counts on top of each bar
    for bar in bars.patches:
        height = bar.get_height()
        plt.text(
            bar.get_x() + bar.get_width() / 2,
            height,
            f'{int(height)}',
            ha='center',
            va='bottom',
            fontsize=12,
            color='black'
        )
    
    plt.xticks(rotation=45, ha='right', fontsize=14)
    plt.yticks(fontsize=14)
    plt.title(title, fontsize=21, pad=10)
    plt.xlabel("NER Tags", fontsize=18)
    plt.ylabel("Count", fontsize=18)
    plt.tight_layout()
    
    # Save Plot
    file_name = f"RandomForest_{title.replace(' ', '_').replace(':', '')}"
    plt.savefig(f"{PLOT_DIR}/{file_name}.png", dpi=300, bbox_inches='tight')
    plt.show()
    plt.close()

# Generate NER tag distribution plots for all dataset splits.
plot_ner_distribution(y_train, "NER Tag Distribution in Training Set")
plot_ner_distribution(y_val, "NER Tag Distribution in Validation Set")
plot_ner_distribution(y_test, "NER Tag Distribution in Test Set")

# ## Step 4: Vectorization
vectorizer = DictVectorizer(sparse=True)
# Keep track of sequence lengths before flattening
train_seq_lengths = [len(seq) for seq in X_train_feats]
X_train_flat = [item for sublist in X_train_feats for item in sublist]
X_train_vec = vectorizer.fit_transform(X_train_flat)
del X_train_flat
gc.collect()

X_val_vec = vectorizer.transform([item for sublist in X_val_feats for item in sublist])
X_test_vec = vectorizer.transform([item for sublist in X_test_feats for item in sublist])

# ### Flatten and Encode Labels
y_train_flat = [item for sublist in y_train for item in sublist]
y_val_flat = [item for sublist in y_val for item in sublist]
y_test_flat = [item for sublist in y_test for item in sublist]

label_encoder = LabelEncoder()
y_train_encoded = label_encoder.fit_transform(y_train_flat)
y_val_encoded = label_encoder.transform(y_val_flat)
y_test_encoded = label_encoder.transform(y_test_flat)

del y_train_flat, y_val_flat, y_test_flat
gc.collect()

# ## Step 5: Train Random Forest with Class Weighting
rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1,
    max_depth=30,
    min_samples_split=5,
    class_weight='balanced'
)
rf.fit(X_train_vec, y_train_encoded)

# ## Step 6-9: Evaluation with Confusion Matrix for All Sets
def evaluate_model(X, y_true, split_name, labels):
    y_pred = rf.predict(X)
    print(f"{split_name} Classification Report:")
    report = classification_report(
        y_true, 
        y_pred, 
        target_names=labels,
        zero_division=0,
        output_dict=True
    )
    print(classification_report(y_true, y_pred, target_names=labels, zero_division=0))
    
    # Save Classification Report as Excel
    report_df = pd.DataFrame(report).transpose()
    report_file_name = f"{split_name.replace(' ', '_')}_Classification_Report_RandomForest.xlsx"
    report_df.to_excel(f"{PLOT_DIR}/{report_file_name}", sheet_name=split_name)
    
    # Classification Report Bar Chart with Statistics and Excel Export
    classes = [c for c in report.keys() if c not in ['accuracy', 'macro avg', 'weighted avg']]
    precision = [report[c]['precision'] for c in classes]
    recall = [report[c]['recall'] for c in classes]
    f1 = [report[c]['f1-score'] for c in classes]
    
    # Print Classification Report Statistics
    print(f"\nClassification Report Statistics for {split_name}:")
    for cls, p, r, f in zip(classes, precision, recall, f1):
        print(f"{cls}: Precision = {p:.2f}, Recall = {r:.2f}, F1-Score = {f:.2f}")
    
    # Export Classification Report Metrics to Excel
    report_metrics = {
        'Class': classes,
        'Precision': precision,
        'Recall': recall,
        'F1-Score': f1
    }
    report_metrics_df = pd.DataFrame(report_metrics)
    report_metrics_file_name = f"RandomForest_{split_name.replace(' ', '_')}_Classification_Metrics.xlsx"
    report_metrics_df.to_excel(f"{PLOT_DIR}/{report_metrics_file_name}", index=False)
    
    # Plot Classification Report as Bar Chart
    x = np.arange(len(classes))
    width = 0.25
    plt.figure(figsize=(14, 6))
    bars_precision = plt.bar(x - width, precision, width, label='Precision', color='#FF6F61', edgecolor='black', linewidth=1.5)
    bars_recall = plt.bar(x, recall, width, label='Recall', color='#6B5B95', edgecolor='black', linewidth=1.5)
    bars_f1 = plt.bar(x + width, f1, width, label='F1-Score', color='#88B04B', edgecolor='black', linewidth=1.5)
    
    # Add values on top of each bar
    for bars, values in [(bars_precision, precision), (bars_recall, recall), (bars_f1, f1)]:
        for bar, value in zip(bars, values):
            height = bar.get_height()
            plt.text(
                bar.get_x() + bar.get_width() / 2,
                height,
                f'{value:.2f}',
                ha='center',
                va='bottom',
                fontsize=10,
                color='black'
            )
    
    plt.xticks(x, classes, rotation=45, ha='right', fontsize=14)
    plt.yticks(fontsize=14)
    plt.xlabel('Classes', fontsize=18)
    plt.ylabel('Score', fontsize=18)
    report_title = f'Classification Report Metrics - {split_name}'
    plt.title(report_title, fontsize=21)
    plt.legend(fontsize=12)
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    
    # Save Classification Report Plot
    report_file_name = f"RandomForest_{report_title.replace(' ', '_').replace('-', '_')}"
    plt.savefig(f"{PLOT_DIR}/{report_file_name}.png", dpi=300, bbox_inches='tight')
    plt.show()
    plt.close()
    
    # Precision-Recall Curve for Key Classes with Statistics and Excel Export
    plt.figure(figsize=(10, 6))
    pr_data = {}
    for label in ['B-PER', 'I-PER', 'B-LOC', 'O']:
        label_idx = label_encoder.transform([label])[0]
        y_true_binary = (y_true == label_idx).astype(int)
        y_score = rf.predict_proba(X)[:, label_idx]
        precision, recall, _ = precision_recall_curve(y_true_binary, y_score)
        pr_auc = auc(recall, precision)
        
        # Store precision-recall data
        pr_data[label] = {'Recall': recall, 'Precision': precision, 'AUC': pr_auc}
        
        plt.plot(recall, precision, label=f'{label} (AUC = {pr_auc:.2f})', linewidth=2.5)
    
    # Print Precision-Recall Statistics
    print(f"\nPrecision-Recall Statistics for {split_name}:")
    for label, data in pr_data.items():
        print(f"{label} - AUC: {data['AUC']:.2f}")
        print(f"  Sample Recall Points: {data['Recall'][:5]}...")
        print(f"  Sample Precision Points: {data['Precision'][:5]}...")
    
    # Export Precision-Recall Data to Excel
    with pd.ExcelWriter(f"{PLOT_DIR}/RandomForest_Precision_Recall_{split_name.replace(' ', '_')}.xlsx") as writer:
        for label, data in pr_data.items():
            pr_df = pd.DataFrame({'Recall': data['Recall'], 'Precision': data['Precision']})
            pr_df['AUC'] = data['AUC']
            pr_df.to_excel(writer, sheet_name=label)
    
    plt.xticks(fontsize=14)
    plt.yticks(fontsize=14)
    plt.xlabel('Recall', fontsize=18)
    plt.ylabel('Precision', fontsize=18)
    pr_title = f'Precision-Recall Curve - {split_name}'
    plt.title(pr_title, fontsize=21)
    plt.legend(loc='best', fontsize=12)
    plt.grid(True)
    plt.tight_layout()
    
    # Save Precision-Recall Plot
    pr_file_name = f"RandomForest_{pr_title.replace(' ', '_').replace('-', '_')}"
    plt.savefig(f"{PLOT_DIR}/{pr_file_name}.png", dpi=300, bbox_inches='tight')
    plt.show()
    plt.close()
    
    # Confusion Matrix with Excel Export
    cm = confusion_matrix(y_true, y_pred, labels=range(len(labels)))
    print(f"\nConfusion Matrix Statistics for {split_name}:")
    print(cm)
    
    # Export Confusion Matrix to Excel
    cm_df = pd.DataFrame(cm, index=labels, columns=labels)
    cm_file_name = f"RandomForest_Confusion_Matrix_{split_name.replace(' ', '_')}.xlsx"
    cm_df.to_excel(f"{PLOT_DIR}/{cm_file_name}")
    
    plt.figure(figsize=(12, 10))
    sns.heatmap(
        cm, 
        annot=True,
        fmt='d',
        xticklabels=labels,
        yticklabels=labels,
        cmap='RdYlBu',
        cbar_kws={'label': 'Count'},
        linewidths=0.5,
        linecolor='gray'
    )
    plt.xticks(rotation=45, ha='right', fontsize=14)
    plt.yticks(rotation=0, fontsize=14)
    plt.xlabel("Predicted", fontsize=18)
    plt.ylabel("True", fontsize=18)
    cm_title = f"Confusion Matrix - {split_name}"
    plt.title(cm_title, fontsize=21, pad=10)
    plt.tight_layout()
    
    # Save Confusion Matrix Plot
    cm_file_name = f"RandomForest_{cm_title.replace(' ', '_').replace('-', '_')}"
    plt.savefig(f"{PLOT_DIR}/{cm_file_name}.png", dpi=300, bbox_inches='tight')
    plt.show()
    plt.close()
    
    return y_pred, report

# Evaluate the model on all dataset splits.
y_train_pred, train_report = evaluate_model(X_train_vec, y_train_encoded, "Training Set", label_encoder.classes_)
y_val_pred, val_report = evaluate_model(X_val_vec, y_val_encoded, "Validation Set", label_encoder.classes_)
y_test_pred, test_report = evaluate_model(X_test_vec, y_test_encoded, "Test Set", label_encoder.classes_)

# ## Step 10: Transition Weights Visualization with Statistics and Excel Export
def plot_transition_weights(rf, label_encoder, X_data, train_seq_lengths, top_n=20):
    # Predict labels for the training data to estimate transitions
    y_pred = rf.predict(X_data)
    labels = label_encoder.classes_
    
    # Reconstruct the predicted sequences using the original sequence lengths
    pred_sequences = []
    start_idx = 0
    for seq_len in train_seq_lengths:
        end_idx = start_idx + seq_len
        pred_seq = y_pred[start_idx:end_idx]
        pred_sequences.append(pred_seq)
        start_idx = end_idx
    
    # Compute transition counts between consecutive labels
    transition_counts = Counter()
    total_transitions = 0
    
    # Iterate over each sequence of predictions
    for seq in pred_sequences:
        for i in range(len(seq) - 1):
            from_label = labels[seq[i]]
            to_label = labels[seq[i + 1]]
            transition_counts[(from_label, to_label)] += 1
            total_transitions += 1
    
    # Compute transition probabilities (pseudo-weights)
    transitions = {}
    for (from_label, to_label), count in transition_counts.items():
        transitions[(from_label, to_label)] = count / total_transitions if total_transitions > 0 else 0
    
    # Sort transitions by probability
    sorted_trans = sorted(transitions.items(), key=lambda x: abs(x[1]), reverse=True)[:top_n]
    from_labels, to_labels = zip(*[t[0] for t in sorted_trans])
    weights = [t[1] for t in sorted_trans]
    
    # Print Transition Weights Statistics
    print("\nTransition Weights Statistics:")
    transition_dict = {f"{f} → {t}": w for (f, t), w in zip(zip(from_labels, to_labels), weights)}
    for transition, weight in transition_dict.items():
        print(f"{transition}: {weight:.4f}")
    
    # Export to Excel
    transition_df = pd.DataFrame.from_dict(transition_dict, orient='index', columns=['Weight'])
    transition_df.index.name = 'Transition'
    transition_df.to_excel(f"{PLOT_DIR}/RandomForest_Transition_Weights.xlsx")
    
    # Plot
    plt.figure(figsize=(12, 6))
    bars = sns.barplot(
        x=[f"{f} → {t}" for f, t in zip(from_labels, to_labels)],
        y=weights,
        hue=[f"{f} → {t}" for f, t in zip(from_labels, to_labels)],
        palette="magma",
        legend=False,
        edgecolor="black",
        linewidth=1.5
    )
    
    # Add weights on top of each bar
    for bar in bars.patches:
        height = bar.get_height()
        plt.text(
            bar.get_x() + bar.get_width() / 2,
            height if height >= 0 else height - 0.5,  # Adjust position for negative weights
            f'{height:.2f}',
            ha='center',
            va='bottom' if height >= 0 else 'top',
            fontsize=10,
            color='black'
        )
    
    plt.xticks(rotation=45, ha='right', fontsize=14)
    plt.yticks(fontsize=14)
    plt.title("Top Transition Weights in Random Forest", fontsize=21)
    plt.xlabel("Transitions", fontsize=18)
    plt.ylabel("Weight", fontsize=18)
    plt.tight_layout()
    
    # Save Plot
    file_name = f"Top_Transition_Weights_in_Random_Forest_RandomForest"
    plt.savefig(f"{PLOT_DIR}/{file_name}.png", dpi=300, bbox_inches='tight')
    plt.show()
    plt.close()

# Generate transition weights plot.
plot_transition_weights(rf, label_encoder, X_train_vec, train_seq_lengths)

# ## Step 11: Feature Importance Plot with Statistics and Excel Export
def plot_feature_importance(rf, vectorizer, top_n=20):
    # Extract feature importances
    importances = rf.feature_importances_
    feature_names = vectorizer.feature_names_
    
    # Create a dictionary of all feature importances
    all_feature_importances = {}
    for feat, importance in zip(feature_names, importances):
        feature_key = f"{feat}"
        all_feature_importances[feature_key] = importance
    
    # Export all feature importances to Excel
    all_feature_importances_df = pd.DataFrame.from_dict(all_feature_importances, orient='index', columns=['Importance'])
    all_feature_importances_df.index.name = 'Feature'
    all_feature_importances_df = all_feature_importances_df.sort_values(by='Importance', key=abs, ascending=False)  # Sort by absolute importance
    all_feature_importances_df.to_excel(f"{PLOT_DIR}/RandomForest_All_Feature_Importance.xlsx")
    
    # Select top N features by absolute importance for plotting
    sorted_features = sorted(zip(feature_names, importances), key=lambda x: abs(x[1]), reverse=True)[:top_n]
    feature_labels = [feat for feat, _ in sorted_features]
    weights = [importance for _, importance in sorted_features]
    
    # Print Feature Importance Statistics (Top N)
    print(f"\nTop {top_n} Feature Importance Statistics:")
    top_feature_importances_dict = {f"{feat}": importance for feat, importance in sorted_features}
    for feature_label, importance in top_feature_importances_dict.items():
        print(f"{feature_label}: {importance:.4f}")
    
    # Export Top N feature importances to Excel
    top_feature_importances_df = pd.DataFrame.from_dict(top_feature_importances_dict, orient='index', columns=['Importance'])
    top_feature_importances_df.index.name = 'Feature'
    top_feature_importances_df.to_excel(f"{PLOT_DIR}/RandomForest_Top_{top_n}_Feature_Importance.xlsx")
    
    # Plot Top N Feature Importances
    plt.figure(figsize=(12, 6))
    bars = sns.barplot(
        x=feature_labels,
        y=weights,
        hue=feature_labels,
        palette="magma",
        legend=False,
        edgecolor="black",
        linewidth=1.5
    )
    
    # Add importance values on top of each bar
    for bar in bars.patches:
        height = bar.get_height()
        plt.text(
            bar.get_x() + bar.get_width() / 2,
            height if height >= 0 else height - 0.5,  # Adjust position for negative weights
            f'{height:.2f}',
            ha='center',
            va='bottom' if height >= 0 else 'top',
            fontsize=10,
            color='black'
        )
    
    plt.xticks(rotation=45, ha='right', fontsize=14)
    plt.yticks(fontsize=14)
    title = f"Top {top_n} Feature Importances in Random Forest"
    plt.title(title, fontsize=21)
    plt.xlabel("Feature", fontsize=18)  # Adjusted to remove "→ Label" since RF doesn't have labels in feature importance
    plt.ylabel("Importance", fontsize=18)  # Changed "Weight" to "Importance" for RF context
    plt.tight_layout()
    
    # Save Plot
    file_name = f"RandomForest_{title.replace(' ', '_')}"
    plt.savefig(f"{PLOT_DIR}/{file_name}.png", dpi=300, bbox_inches='tight')
    plt.show()
    plt.close()

# Generate feature importance plot.
plot_feature_importance(rf, vectorizer)

# ## Step 12: F1-Score Comparison Plot with Statistics and Excel Export
def plot_f1_comparison(train_report, val_report, test_report, title):
    classes = [c for c in val_report.keys() if c not in ['accuracy', 'macro avg', 'weighted avg']]
    train_f1 = [train_report[c]['f1-score'] for c in classes]
    val_f1 = [val_report[c]['f1-score'] for c in classes]
    test_f1 = [test_report[c]['f1-score'] for c in classes]
    
    # Print F1-Score Statistics
    print(f"\nF1-Score Comparison Statistics for {title}:")
    for cls, t_f1, v_f1, te_f1 in zip(classes, train_f1, val_f1, test_f1):
        print(f"{cls}: Training F1 = {t_f1:.2f}, Validation F1 = {v_f1:.2f}, Test F1 = {te_f1:.2f}")
    
    # Export to Excel
    f1_data = {
        'Class': classes,
        'Training F1': train_f1,
        'Validation F1': val_f1,
        'Test F1': test_f1
    }
    f1_df = pd.DataFrame(f1_data)
    f1_file_name = f"RandomForest_{title.replace(' ', '_').replace(':', '')}_F1_Scores.xlsx"
    f1_df.to_excel(f"{PLOT_DIR}/{f1_file_name}", index=False)
    
    # Plot
    x = np.arange(len(classes))
    width = 0.25
    plt.figure(figsize=(14, 6))
    bars_train = plt.bar(x - width, train_f1, width, label='Training', color='#FF6F61', edgecolor='black', linewidth=1.5)
    bars_val = plt.bar(x, val_f1, width, label='Validation', color='#6B5B95', edgecolor='black', linewidth=1.5)
    bars_test = plt.bar(x + width, test_f1, width, label='Test', color='#88B04B', edgecolor='black', linewidth=1.5)
    
    # Add F1-scores on top of each bar
    for bars, f1_scores in [(bars_train, train_f1), (bars_val, val_f1), (bars_test, test_f1)]:
        for bar, f1 in zip(bars, f1_scores):
            height = bar.get_height()
            plt.text(
                bar.get_x() + bar.get_width() / 2,
                height,
                f'{f1:.2f}',
                ha='center',
                va='bottom',
                fontsize=10,
                color='black'
            )
    
    plt.xticks(x, classes, rotation=45, ha='right', fontsize=14)
    plt.yticks(fontsize=14)
    plt.xlabel('Classes', fontsize=18)
    plt.ylabel('F1-Score', fontsize=18)
    plt.title(title, fontsize=21)
    plt.legend(fontsize=12)
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    
    # Save Plot
    file_name = f"RandomForest_{title.replace(' ', '_').replace(':', '')}"
    plt.savefig(f"{PLOT_DIR}/{file_name}.png", dpi=300, bbox_inches='tight')
    plt.show()
    plt.close()

# Generate F1-score comparison plot.
plot_f1_comparison(train_report, val_report, test_report, "F1-Score Comparison: Training vs Validation vs Test")

# ## Cleanup
del X_train_vec, X_val_vec, X_test_vec
gc.collect()
```

CRF

```python
# ## Imports
# Import necessary libraries for data processing, modeling, visualization, and Excel file handling.
import matplotlib.pyplot as plt  # For creating plots
import seaborn as sns           # For enhanced visualizations with better aesthetics
import numpy as np              # For numerical operations and array handling
from datasets import load_dataset  # To load the CoNLL-2003 dataset
from sklearn_crfsuite import CRF   # Conditional Random Field model for sequence labeling
from sklearn.metrics import classification_report, confusion_matrix, precision_recall_curve, auc  # Evaluation metrics
from collections import Counter    # For counting NER tag occurrences
import os                          # For directory handling and file saving
import pandas as pd                # For saving classification reports as Excel files

# ## Initial Setup
# Configure Seaborn for consistent and visually appealing plots.
sns.set(style="whitegrid", palette="muted")  # Set white grid background and muted color palette

# ## Plot Directory Setup
# Define a fixed directory for saving all plots and reports.
PLOT_DIR = "plots2/CRF"              # Directory path for CRF plots and reports
os.makedirs(PLOT_DIR, exist_ok=True)  # Create directory if it doesn't exist

# ## Step 1: Load CoNLL-2003 Dataset
# Load the CoNLL-2003 dataset and extract NER tag names.
conll2003 = load_dataset("conll2003")  # Load dataset without streaming (full load into memory)
label_names = conll2003["train"].features["ner_tags"].feature.names  # Get list of NER tag names (e.g., "O", "B-PER")

# ## Step 2: Prepare Data for CRF
# Define a function to prepare sentences and labels for CRF training.
def prepare_data_for_crf(dataset):
    sentences, labels = [], []
    for example in dataset:
        words = example['tokens']
        ner_tags = [label_names[tag] for tag in example['ner_tags']]
        sentences.append(words)
        labels.append(ner_tags)
    return sentences, labels

# Prepare training, validation, and test sets for CRF.
X_train, y_train = prepare_data_for_crf(conll2003['train'])
X_val, y_val = prepare_data_for_crf(conll2003['validation'])
X_test, y_test = prepare_data_for_crf(conll2003['test'])

# ## Step 3: Enhanced NER Tag Distribution Plot with Statistics and Excel Export
def plot_ner_distribution(y_data, title):
    label_counts = Counter(tag for seq in y_data for tag in seq)
    
    # Print the statistics
    print(f"\nNER Tag Distribution Statistics for {title}:")
    for tag, count in label_counts.items():
        print(f"{tag}: {count}")
    
    # Export to Excel
    counts_df = pd.DataFrame.from_dict(label_counts, orient='index', columns=['Count'])
    counts_df.index.name = 'NER Tag'
    counts_file_name = f"CRF_{title.replace(' ', '_').replace(':', '')}_Counts.xlsx"
    counts_df.to_excel(f"{PLOT_DIR}/{counts_file_name}")
    
    # Create the bar plot
    plt.figure(figsize=(12, 6))
    bars = sns.barplot(
        x=list(label_counts.keys()),
        y=list(label_counts.values()),
        hue=list(label_counts.keys()),
        palette="viridis",
        legend=False,
        edgecolor="black",
        linewidth=1.5
    )
    
    # Add counts on top of each bar
    for bar in bars.patches:
        height = bar.get_height()
        plt.text(
            bar.get_x() + bar.get_width() / 2,
            height,
            f'{int(height)}',
            ha='center',
            va='bottom',
            fontsize=12,
            color='black'
        )
    
    plt.xticks(rotation=45, ha='right', fontsize=14)
    plt.yticks(fontsize=14)
    plt.title(title, fontsize=21, pad=10)
    plt.xlabel("NER Tags", fontsize=18)
    plt.ylabel("Count", fontsize=18)
    plt.tight_layout()
    
    # Save Plot
    file_name = f"{title.replace(' ', '_').replace(':', '')}_CRF"
    plt.savefig(f"{PLOT_DIR}/{file_name}.png", dpi=300, bbox_inches='tight')
    plt.show()
    plt.close()

# Generate NER tag distribution plots for all dataset splits.
plot_ner_distribution(y_train, "NER Tag Distribution in Training Set")
plot_ner_distribution(y_val, "NER Tag Distribution in Validation Set")
plot_ner_distribution(y_test, "NER Tag Distribution in Test Set")

# ## Step 4: Feature Extraction for CRF
def word2features(sent, i):
    word = sent[i]
    features = {
        'bias': 1.0,
        'word.lower': word.lower(),
        'word[-3:]': word[-3:],
        'word[-2:]': word[-2:],
        'word.isupper': word.isupper(),
        'word.istitle': word.istitle(),
        'word.isdigit': word.isdigit()
    }
    if i > 0:
        word1 = sent[i-1]
        features.update({
            'prev_word.lower': word1.lower(),
            'prev_word.istitle': word1.istitle(),
            'prev_word.isupper': word1.isupper()
        })
    else:
        features['BOS'] = True
    if i < len(sent)-1:
        word1 = sent[i+1]
        features.update({
            'next_word.lower': word1.lower(),
            'next_word.istitle': word1.istitle(),
            'next_word.isupper': word1.isupper()
        })
    else:
        features['EOS'] = True
    return features

def extract_features(sentences):
    return [[word2features(sent, i) for i in range(len(sent))] for sent in sentences]

# Extract features for training, validation, and test sets.
X_train_feats = extract_features(X_train)
X_val_feats = extract_features(X_val)
X_test_feats = extract_features(X_test)

# ## Step 5: Train CRF Model
crf = CRF(
    algorithm='lbfgs',
    c1=0.1,
    c2=0.1,
    max_iterations=100,
    all_possible_transitions=True
)
crf.fit(X_train_feats, y_train)

# ## Step 6: Evaluation with All Plots
def evaluate_model(X, y_true, split_name, labels):
    y_pred = crf.predict(X)
    y_true_flat = [item for sublist in y_true for item in sublist]
    y_pred_flat = [item for sublist in y_pred for item in sublist]
    
    print(f"{split_name} Classification Report:")
    report = classification_report(y_true_flat, y_pred_flat, target_names=labels, output_dict=True)
    print(classification_report(y_true_flat, y_pred_flat, target_names=labels))
    
    # Save Classification Report as Excel
    report_df = pd.DataFrame(report).transpose()
    report_file_name = f"{split_name.replace(' ', '_')}_Classification_Report_CRF.xlsx"
    report_df.to_excel(f"{PLOT_DIR}/{report_file_name}", sheet_name=split_name)
    
    # Classification Report Bar Chart with Statistics and Excel Export
    classes = [c for c in report.keys() if c not in ['accuracy', 'macro avg', 'weighted avg']]
    precision = [report[c]['precision'] for c in classes]
    recall = [report[c]['recall'] for c in classes]
    f1 = [report[c]['f1-score'] for c in classes]
    
    # Print Classification Report Statistics
    print(f"\nClassification Report Statistics for {split_name}:")
    for cls, p, r, f in zip(classes, precision, recall, f1):
        print(f"{cls}: Precision = {p:.2f}, Recall = {r:.2f}, F1-Score = {f:.2f}")
    
    # Export Classification Report Metrics to Excel
    report_metrics = {
        'Class': classes,
        'Precision': precision,
        'Recall': recall,
        'F1-Score': f1
    }
    report_metrics_df = pd.DataFrame(report_metrics)
    report_metrics_file_name = f"CRF_{split_name.replace(' ', '_')}_Classification_Metrics.xlsx"
    report_metrics_df.to_excel(f"{PLOT_DIR}/{report_metrics_file_name}", index=False)
    
    # Plot Classification Report as Bar Chart
    x = np.arange(len(classes))
    width = 0.25
    plt.figure(figsize=(14, 6))
    bars_precision = plt.bar(x - width, precision, width, label='Precision', color='#FF6F61', edgecolor='black', linewidth=1.5)
    bars_recall = plt.bar(x, recall, width, label='Recall', color='#6B5B95', edgecolor='black', linewidth=1.5)
    bars_f1 = plt.bar(x + width, f1, width, label='F1-Score', color='#88B04B', edgecolor='black', linewidth=1.5)
    
    # Add values on top of each bar
    for bars, values in [(bars_precision, precision), (bars_recall, recall), (bars_f1, f1)]:
        for bar, value in zip(bars, values):
            height = bar.get_height()
            plt.text(
                bar.get_x() + bar.get_width() / 2,
                height,
                f'{value:.2f}',
                ha='center',
                va='bottom',
                fontsize=10,
                color='black'
            )
    
    plt.xticks(x, classes, rotation=45, ha='right', fontsize=14)
    plt.yticks(fontsize=14)
    plt.xlabel('Classes', fontsize=18)
    plt.ylabel('Score', fontsize=18)
    report_title = f'Classification Report Metrics - {split_name}'
    plt.title(report_title, fontsize=21)
    plt.legend(fontsize=12)
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    
    # Save Classification Report Plot
    report_file_name = f"CRF_{report_title.replace(' ', '_').replace('-', '_')}"
    plt.savefig(f"{PLOT_DIR}/{report_file_name}.png", dpi=300, bbox_inches='tight')
    plt.show()
    plt.close()
    
    # Precision-Recall Curve with Statistics and Excel Export
    plt.figure(figsize=(10, 6))
    pr_data = {}
    plotted_curves = False
    for label in ['B-PER', 'I-PER', 'B-LOC', 'O']:
        try:
            label_idx = labels.index(label)
            y_true_binary = [1 if tag == label else 0 for tag in y_true_flat]
            marginals = [prob[label] for seq in crf.predict_marginals(X) for prob in seq]
            y_score = marginals[:len(y_true_binary)]
            precision, recall, _ = precision_recall_curve(y_true_binary, y_score)
            pr_auc = auc(recall, precision)
            
            # Store precision-recall data
            pr_data[label] = {'Recall': recall, 'Precision': precision, 'AUC': pr_auc}
            
            plt.plot(recall, precision, label=f'{label} (AUC = {pr_auc:.2f})', linewidth=2.5)
            plotted_curves = True
        except (ValueError, KeyError):
            continue
    
    # Print Precision-Recall Statistics
    print(f"\nPrecision-Recall Statistics for {split_name}:")
    for label, data in pr_data.items():
        print(f"{label} - AUC: {data['AUC']:.2f}")
        print(f"  Sample Recall Points: {data['Recall'][:5]}...")
        print(f"  Sample Precision Points: {data['Precision'][:5]}...")
    
    # Export Precision-Recall Data to Excel
    with pd.ExcelWriter(f"{PLOT_DIR}/CRF_Precision_Recall_{split_name.replace(' ', '_')}.xlsx") as writer:
        for label, data in pr_data.items():
            pr_df = pd.DataFrame({'Recall': data['Recall'], 'Precision': data['Precision']})
            pr_df['AUC'] = data['AUC']
            pr_df.to_excel(writer, sheet_name=label)
    
    plt.xticks(fontsize=14)
    plt.yticks(fontsize=14)
    plt.xlabel('Recall', fontsize=18)
    plt.ylabel('Precision', fontsize=18)
    pr_title = f'Precision-Recall Curve - {split_name}'
    plt.title(pr_title, fontsize=21)
    if plotted_curves:
        plt.legend(loc='best', fontsize=12)
    else:
        plt.text(0.5, 0.5, 'No valid curves plotted', ha='center', va='center', fontsize=12)
    plt.grid(True)
    plt.tight_layout()
    
    # Save Precision-Recall Plot
    pr_file_name = f"{pr_title.replace(' ', '_').replace('-', '_')}_CRF"
    plt.savefig(f"{PLOT_DIR}/{pr_file_name}.png", dpi=300, bbox_inches='tight')
    plt.show()
    plt.close()
    
    # Confusion Matrix with Statistics and Excel Export
    cm = confusion_matrix(y_true_flat, y_pred_flat, labels=labels)
    print(f"\nConfusion Matrix Statistics for {split_name}:")
    print(cm)
    
    # Export Confusion Matrix to Excel
    cm_df = pd.DataFrame(cm, index=labels, columns=labels)
    cm_file_name = f"CRF_Confusion_Matrix_{split_name.replace(' ', '_')}.xlsx"
    cm_df.to_excel(f"{PLOT_DIR}/{cm_file_name}")
    
    plt.figure(figsize=(12, 10))
    sns.heatmap(
        cm, 
        annot=True,
        fmt='d',
        xticklabels=labels,
        yticklabels=labels,
        cmap='RdYlBu',
        cbar_kws={'label': 'Count'},
        linewidths=0.5,
        linecolor='gray'
    )
    plt.xticks(rotation=45, ha='right', fontsize=14)
    plt.yticks(rotation=0, fontsize=14)
    plt.xlabel("Predicted", fontsize=18)
    plt.ylabel("True", fontsize=18)
    cm_title = f"Confusion Matrix - {split_name}"
    plt.title(cm_title, fontsize=21, pad=10)
    plt.tight_layout()
    
    # Save Confusion Matrix Plot
    cm_file_name = f"{cm_title.replace(' ', '_').replace('-', '_')}_CRF"
    plt.savefig(f"{PLOT_DIR}/{cm_file_name}.png", dpi=300, bbox_inches='tight')
    plt.show()
    plt.close()
    
    return y_pred, report

# Evaluate the model on all dataset splits.
y_train_pred, train_report = evaluate_model(X_train_feats, y_train, "Training Set", crf.classes_)
y_val_pred, val_report = evaluate_model(X_val_feats, y_val, "Validation Set", crf.classes_)
y_test_pred, test_report = evaluate_model(X_test_feats, y_test, "Test Set", crf.classes_)

# ## Step 7: Transition Weights Visualization with Statistics and Excel Export
def plot_transition_weights(crf, top_n=20):
    transitions = crf.transition_features_
    sorted_trans = sorted(transitions.items(), key=lambda x: abs(x[1]), reverse=True)[:top_n]
    from_labels, to_labels = zip(*[t[0] for t in sorted_trans])
    weights = [t[1] for t in sorted_trans]
    
    # Print Transition Weights Statistics
    print("\nTransition Weights Statistics:")
    transition_dict = {f"{f} → {t}": w for (f, t), w in zip(zip(from_labels, to_labels), weights)}
    for transition, weight in transition_dict.items():
        print(f"{transition}: {weight:.4f}")
    
    # Export to Excel
    transition_df = pd.DataFrame.from_dict(transition_dict, orient='index', columns=['Weight'])
    transition_df.index.name = 'Transition'
    transition_df.to_excel(f"{PLOT_DIR}/CRF_Transition_Weights.xlsx")
    
    # Plot
    plt.figure(figsize=(12, 6))
    bars = sns.barplot(
        x=[f"{f} → {t}" for f, t in zip(from_labels, to_labels)],
        y=weights,
        hue=[f"{f} → {t}" for f, t in zip(from_labels, to_labels)],
        palette="magma",
        legend=False,
        edgecolor="black",
        linewidth=1.5
    )
    
    # Add weights on top of each bar
    for bar in bars.patches:
        height = bar.get_height()
        plt.text(
            bar.get_x() + bar.get_width() / 2,
            height if height >= 0 else height - 0.5,  # Adjust position for negative weights
            f'{height:.2f}',
            ha='center',
            va='bottom' if height >= 0 else 'top',
            fontsize=10,
            color='black'
        )
    
    plt.xticks(rotation=45, ha='right', fontsize=14)
    plt.yticks(fontsize=14)
    plt.title("Top Transition Weights in CRF", fontsize=21)
    plt.xlabel("Transitions", fontsize=18)
    plt.ylabel("Weight", fontsize=18)
    plt.tight_layout()
    
    # Save Plot
    file_name = f"Top_Transition_Weights_in_CRF_CRF"
    plt.savefig(f"{PLOT_DIR}/{file_name}.png", dpi=300, bbox_inches='tight')
    plt.show()
    plt.close()

# Generate transition weights plot.
plot_transition_weights(crf)

# ## Step 8: Feature Importance Plot with Statistics and Excel Export
def plot_feature_importance(crf, top_n=20):
    # Extract state features (feature weights for label predictions)
    state_features = crf.state_features_
    
    # Create a dictionary of all feature weights
    all_feature_weights = {}
    for (feat, label), weight in state_features.items():
        feature_key = f"{feat} → {label}"
        all_feature_weights[feature_key] = weight
    
    # Export all feature weights to Excel
    all_feature_weights_df = pd.DataFrame.from_dict(all_feature_weights, orient='index', columns=['Weight'])
    all_feature_weights_df.index.name = 'Feature_Label'
    all_feature_weights_df = all_feature_weights_df.sort_values(by='Weight', key=abs, ascending=False)  # Sort by absolute weight
    all_feature_weights_df.to_excel(f"{PLOT_DIR}/CRF_All_Feature_Importance.xlsx")
    
    # Select top N features by absolute weight for plotting
    sorted_features = sorted(state_features.items(), key=lambda x: abs(x[1]), reverse=True)[:top_n]
    feature_labels = [f"{feat} → {label}" for (feat, label), _ in sorted_features]
    weights = [weight for _, weight in sorted_features]
    
    # Print Feature Importance Statistics (Top N)
    print(f"\nTop {top_n} Feature Importance Statistics (State Features):")
    top_feature_weights_dict = {f"{feat} → {label}": weight for (feat, label), weight in sorted_features}
    for feature_label, weight in top_feature_weights_dict.items():
        print(f"{feature_label}: {weight:.4f}")
    
    # Export Top N feature weights to Excel
    top_feature_weights_df = pd.DataFrame.from_dict(top_feature_weights_dict, orient='index', columns=['Weight'])
    top_feature_weights_df.index.name = 'Feature_Label'
    top_feature_weights_df.to_excel(f"{PLOT_DIR}/CRF_Top_{top_n}_Feature_Importance.xlsx")
    
    # Plot Top N Feature Importances
    plt.figure(figsize=(12, 6))
    bars = sns.barplot(
        x=feature_labels,
        y=weights,
        hue=feature_labels,
        palette="magma",
        legend=False,
        edgecolor="black",
        linewidth=1.5
    )
    
    # Add weights on top of each bar
    for bar in bars.patches:
        height = bar.get_height()
        plt.text(
            bar.get_x() + bar.get_width() / 2,
            height if height >= 0 else height - 0.5,  # Adjust position for negative weights
            f'{height:.2f}',
            ha='center',
            va='bottom' if height >= 0 else 'top',
            fontsize=10,
            color='black'
        )
    
    plt.xticks(rotation=45, ha='right', fontsize=14)
    plt.yticks(fontsize=14)
    title = f"Top {top_n} Feature Importances in CRF"
    plt.title(title, fontsize=21)
    plt.xlabel("Feature → Label", fontsize=18)
    plt.ylabel("Weight", fontsize=18)
    plt.tight_layout()
    
    # Save Plot
    file_name = f"CRF_{title.replace(' ', '_')}"
    plt.savefig(f"{PLOT_DIR}/{file_name}.png", dpi=300, bbox_inches='tight')
    plt.show()
    plt.close()

# Generate feature importance plot.
plot_feature_importance(crf)

# ## Step 9: F1-Score Comparison Plot with Statistics and Excel Export
def plot_f1_comparison(train_report, val_report, test_report, title):
    classes = [c for c in val_report.keys() if c not in ['accuracy', 'macro avg', 'weighted avg']]
    train_f1 = [train_report[c]['f1-score'] for c in classes]
    val_f1 = [val_report[c]['f1-score'] for c in classes]
    test_f1 = [test_report[c]['f1-score'] for c in classes]
    
    # Print F1-Score Statistics
    print(f"\nF1-Score Comparison Statistics for {title}:")
    for cls, t_f1, v_f1, te_f1 in zip(classes, train_f1, val_f1, test_f1):
        print(f"{cls}: Training F1 = {t_f1:.2f}, Validation F1 = {v_f1:.2f}, Test F1 = {te_f1:.2f}")
    
    # Export to Excel
    f1_data = {
        'Class': classes,
        'Training F1': train_f1,
        'Validation F1': val_f1,
        'Test F1': test_f1
    }
    f1_df = pd.DataFrame(f1_data)
    f1_file_name = f"CRF_{title.replace(' ', '_').replace(':', '')}_F1_Scores.xlsx"
    f1_df.to_excel(f"{PLOT_DIR}/{f1_file_name}", index=False)
    
    # Plot
    x = np.arange(len(classes))
    width = 0.25
    plt.figure(figsize=(14, 6))
    bars_train = plt.bar(x - width, train_f1, width, label='Training', color='#FF6F61', edgecolor='black', linewidth=1.5)
    bars_val = plt.bar(x, val_f1, width, label='Validation', color='#6B5B95', edgecolor='black', linewidth=1.5)
    bars_test = plt.bar(x + width, test_f1, width, label='Test', color='#88B04B', edgecolor='black', linewidth=1.5)
    
    # Add F1-scores on top of each bar
    for bars, f1_scores in [(bars_train, train_f1), (bars_val, val_f1), (bars_test, test_f1)]:
        for bar, f1 in zip(bars, f1_scores):
            height = bar.get_height()
            plt.text(
                bar.get_x() + bar.get_width() / 2,
                height,
                f'{f1:.2f}',
                ha='center',
                va='bottom',
                fontsize=10,
                color='black'
            )
    
    plt.xticks(x, classes, rotation=45, ha='right', fontsize=14)
    plt.yticks(fontsize=14)
    plt.xlabel('Classes', fontsize=18)
    plt.ylabel('F1-Score', fontsize=18)
    plt.title(title, fontsize=21)
    plt.legend(fontsize=12)
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    
    # Save Plot
    file_name = f"{title.replace(' ', '_').replace(':', '')}_CRF"
    plt.savefig(f"{PLOT_DIR}/{file_name}.png", dpi=300, bbox_inches='tight')
    plt.show()
    plt.close()

# Generate F1-score comparison plot.
plot_f1_comparison(train_report, val_report, test_report, "F1-Score Comparison: Training vs Validation vs Test")
```
